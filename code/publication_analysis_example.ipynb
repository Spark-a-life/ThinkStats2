# Publication Analysis: Codesign and AI Research

This notebook demonstrates how to analyze publication data to find codesign and AI researcher publications.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from publication_analysis import PublicationAnalyzer

%matplotlib inline
```

## 1. Load Your Publication Data

Your data should be in CSV or JSON format with these columns:
- `title`: Publication title
- `authors`: Author names (comma-separated)
- `year`: Publication year
- `keywords`: Keywords (comma-separated)
- `abstract`: Abstract text (optional)
- `venue`: Conference/journal name (optional)

```python
# Replace 'your_data.csv' with your actual data file
DATA_FILE = 'publications.csv'  # <-- CHANGE THIS

# Initialize the analyzer
analyzer = PublicationAnalyzer(DATA_FILE)

# Preview the data
analyzer.data.head()
```

## 2. Filter Codesign Publications

Search for publications containing codesign-related terms.

```python
# Filter for codesign publications
codesign_pubs = analyzer.filter_codesign_publications()

# Display sample results
print(f"\nSample codesign publications:")
codesign_pubs[['title', 'year', 'authors']].head(10)
```

## 3. Filter AI Publications

Search for publications containing AI-related terms.

```python
# Filter for AI publications
ai_pubs = analyzer.filter_ai_publications()

# Display sample results
print(f"\nSample AI publications:")
ai_pubs[['title', 'year', 'authors']].head(10)
```

## 4. Find Intersection: Codesign + AI

Find publications that combine both codesign and AI.

```python
# Find publications at intersection
intersection = analyzer.find_intersection()

# Display all intersection publications
print(f"\nPublications combining codesign and AI:")
intersection[['title', 'year', 'authors']]
```

## 5. Temporal Analysis

Analyze publication trends over time.

```python
# Analyze codesign publication trends
print("Codesign Publication Trends:")
codesign_yearly, codesign_stats = analyzer.analyze_temporal_trends(codesign_pubs)
print(f"\nStatistics:")
for key, value in codesign_stats.items():
    print(f"  {key}: {value}")

# Analyze AI publication trends
print("\nAI Publication Trends:")
ai_yearly, ai_stats = analyzer.analyze_temporal_trends(ai_pubs)
print(f"\nStatistics:")
for key, value in ai_stats.items():
    print(f"  {key}: {value}")
```

## 6. Visualize Trends

Plot publication trends over time.

```python
# Create visualization
fig = analyzer.plot_publication_trends()
plt.tight_layout()
plt.show()
```

## 7. Top Authors

Identify top authors in each category.

```python
# Top codesign authors
print("Top Codesign Authors:")
top_codesign = analyzer.extract_top_authors(codesign_pubs, top_n=10)
print(top_codesign)

# Top AI authors
print("\nTop AI Authors:")
top_ai = analyzer.extract_top_authors(ai_pubs, top_n=10)
print(top_ai)
```

## 8. Summary Report

Generate a comprehensive summary report.

```python
analyzer.generate_summary_report()
```

## 9. Export Results

Save filtered results to CSV files.

```python
# Export codesign publications
codesign_pubs.to_csv('codesign_publications.csv', index=False)
print(f"Exported {len(codesign_pubs)} codesign publications")

# Export AI publications
ai_pubs.to_csv('ai_publications.csv', index=False)
print(f"Exported {len(ai_pubs)} AI publications")

# Export intersection
intersection.to_csv('codesign_ai_intersection.csv', index=False)
print(f"Exported {len(intersection)} codesign+AI publications")
```
